Args:
- `--result_path` (str) - path to folder containing experiment results (required)
- `--base_repo_path` (str) - path to the repo root with `data/` and reference structures (required)
- `--limit_axis` (bool) - limit plot axis to Del Alamo et. al. limits (default: True)
- `--rerun` (str) - what to regenerate: `missing` (default), `plots`, `csv`, or `all`
- `--experiment_folder_name` (str) - plot only a single experiment (should be an experiment folder name, eg. "col_mask_5")
- `--axis_bounds` (list) - pass a list of x and y axis limits in the format x_low x_high y_low y_high, (ex. `--axis_bounds 0.3 0.7 0.8 0.9`) (axis limit param must be set to True!)
- `--guidelines` (bool) - plot TM-score between IF-OF or IN-AC conf.
- `--opacity` (float) - opacity of the scatterplot points (1 - solid, 0 - transparent)
- `--model` (str) - AF model number
- `--seed` (str) - AF model seed
- `--color_on` (str) - column name to color by
- `--shape_on` (str) - column name to shape points by

```python
import os
from plot_tmscore import combine_plots
# The three experiments you want to combine
# experiments = [
#     "query_mask_15_Seed0", 
#     "query_mask_15_Seed1", 
#     "query_mask_15_Seed3"
# ]

experiments = [
    "depth_5120", 
    "query_mask_15_Seed1", 
    "query_mask_15_Seed2", 
    "query_mask_15_Seed7"
]


# Pick a protein to plot (you could also put this in a for-loop for multiple proteins)

# protein_name = "CGRPR" 

protein_name = "MCT1" 


# Locate the CSVs generated by earlier runs
csv_files = [
    os.path.join(result_path, "plots", "TM_Score", exp, f"{protein_name}.csv")
    for exp in experiments
]
# Combine them into one plot!
combine_plots(
    data_files=csv_files,
    save_file_name=f"combined_{protein_name}.png",
    limit_axis=True,
    # axis_bounds=[0.82, 1.0, 0.82, 1.0],
    axis_bounds=[0.7, 1.0, 0.7, 1.0],
    color_on="experiment",
    legend_title="QM Seeds",
    legend_labels=["No QM", "QM Seed 0", "QM Seed 1", "QM Seed 3"],
    output_dir=os.path.join(result_path, "plots", "TM_Score") # Where to save the plot
)
```
