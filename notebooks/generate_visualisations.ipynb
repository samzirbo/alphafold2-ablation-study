```python
base_drive_path = "content/drive/AlphaFold2 Ablation Study"
result_path = base_drive_path.rstrip("/") + "/04_Results"
output_path = result_path + "/plots/viz"
```

Args:
- `--result_path` (str) - path to folder containing experiment results (required)
- `--output_path` (str) - path to folder for PNG/GIF outputs (required)
- `--base_repo_path` (str) - path to the repo root with `data/` and reference structures (default: repo root)
- `--experiment` (str, one or more) - render only selected experiments (optional; default: all experiments)
- `--protein` (str, one or more) - render only selected proteins (optional; default: all 12 proteins)
- `--gif` (bool) - also generate rocking GIFs (default: PNG only)
- `--gif-frames` (int) - number of GIF frames (default: 40)
- `--force` (bool) - regenerate outputs even if they already exist (default: skip existing)

GIFs are off by default. Omit `--gif` to write PNGs only. Omit `--force` to resume and skip already processed files.
