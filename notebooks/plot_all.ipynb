Args:
- `--result_path` (str) - path to folder containing experiment results (required)
- `--base_repo_path` (str) - path to the repo root with `data/` and reference structures (required)
- `--limit_axis` (bool) - limit plot axis to Del Alamo et. al. limits (default: True)
- `--rerun` (str) - what to regenerate: `missing` (default), `plots`, `csv`, or `all`
- `--experiment_folder_name` (str) - plot only a single experiment (should be an experiment folder name, eg. "col_mask_5")
- `--axis_bounds` (list) - pass a list of x and y axis limits in the format x_low x_high y_low y_high, (ex. `--axis_bounds 0.3 0.7 0.8 0.9`) (axis limit param must be set to True!)
- `--guidelines` (bool) - plot TM-score between IF-OF or IN-AC conf.
- `--opacity` (float) - opacity of the scatterplot points (1 - solid, 0 - transparent)
- `--model` (str) - AF model number
- `--seed` (str) - AF model seed
- `--color_on` (str) - column name to color by
- `--shape_on` (str) - column name to shape points by
