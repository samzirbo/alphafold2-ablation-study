```python
base_drive_path = "/content/drive/MyDrive/AlphaFold2 Ablation Study"
base_repo_path = "/content/alphafold2-ablation-study"
result_path = base_drive_path + "/04_Results/" if base_drive_path[-1] != "/" else base_drive_path + "04_Results/"
```

Args:
- `--result_path` (str) - path to folder containing experiment results (required)
- `--base_repo_path` (str) - path to the repo root with `data/` and reference structures (required)
- `--limit_axis` (bool) - limit plot axis to Del Alamo et. al. limits (default: True)
- `--plot_type` (str) - what to plot: `tm` (default, one plot per experiment), `tm_combined` (overlay several experiments in one plot per protein), or `pca`
- `--rerun` (str) - what to regenerate: `missing` (default), `plots`, `csv`, or `all`
- `--experiment_folder_name` (str) - plot only a single experiment (should be an experiment folder name, eg. "col_mask_5")
- `--axis_bounds` (list) - pass a list of x and y axis limits in the format x_low x_high y_low y_high, (ex. `--axis_bounds 0.3 0.7 0.8 0.9`) (axis limit param must be set to True!)
- `--guidelines` (bool) - plot TM-score between IF-OF or IN-AC conf.
- `--opacity` (float) - opacity of the scatterplot points (1 - solid, 0 - transparent)
- `--model` (str) - AF model number
- `--seed` (str) - AF model seed
- `--color_on` (str) - column name to color by (for `tm_combined`, defaults to `experiment`)
- `--shape_on` (str) - column name to shape points by (e.g. `experiment`)
- `--file_name_prefix` (str) - prefix to the plot file name

`tm_combined` only:
- `--experiments` (list) - experiment folder names to overlay in one plot, eg. `--experiments depth_16 depth_256 depth_1024` (default: all experiments)
- `--combined_label` (str) - sub-folder name under `plots/TM_Score/combined/` for the combined outputs (default: `combined`); use it to keep several groupings side by side

`tm_combined` reuses each experiment's existing per-protein CSV (`plots/TM_Score/<experiment>/<protein>.csv`) when available and only recomputes missing ones. Distinguish experiments by color (default, `--color_on experiment`) or by marker shape (`--shape_on experiment --color_on nseq`, which keeps MSA-depth coloring).
