## `generate_overlay_visualizations.py`

Reference-overlay PNGs (and optional rocking GIFs) for one protein.

**Args:**
- `--result_path` (path, required) — folder containing experiment results (e.g. `.../04_Results`)
- `--output_path` (path, required) — folder for PNG/GIF outputs
- `--base_repo_path` (path) — repo root with `data/` and reference structures (default: repo root)
- `--experiment` (str, one or more, required) — experiment folder names under `result_path` (e.g. `depth_5120 query_mask_15`)
- `--protein` (str, required) — protein name (e.g. `LAT1`)
- `--reference-state` (`state_1` | `state_2` | `both`) — which metadata reference conformation(s) to overlay against (default: `both`)
- `--query-mask` (flag) — highlight residues where the experiment A3M query differs from original input sequence
- `--gif` (flag) — also generate rocking GIFs (default: PNG only)
- `--gif-frames` (int) — number of GIF frames (default: 32)

**Output naming:** `reference_overlay_{reference_slug}_{prediction_stem}.png`

GIFs are off by default — omit `--gif` to write PNGs only. Both scripts process a single protein per run; pass multiple experiments as space-separated values to `--experiment`.


```python
! uv run scripts/generate_overlay_visualizations.py \
    --result_path "{result_path}" \
    --output_path "{output_path}" \
    --base_repo_path "{base_repo_path}" \
    --experiment depth_5120 query_mask_15 query_mask_50 \
    --protein LAT1 \
    --query-mask
```
